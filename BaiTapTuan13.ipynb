```python
#W11A2
def canwinNim(path: str):
    try:
        with open(path, 'r', encoding = 'utf-8') as kiet1:
            n = kiet1.read().strip()
            if not n:
                return False
            da = int(n)
            if da % 4 == 0:
                return False
        return True
    except FileNotFoundError:
        print("Khong co duong dan")
canwinNim(r"C:\Users\Admin\Downloads\kiet1.txt")
```

```python
#W11A9
def grade10(path: str):
    try:
        with open(path, 'r', encoding = 'utf-8') as kiet:
            kiet.readline()
            for line in kiet:
                a = line.strip().split(',')
                if a[5] == '10':
                    ketqua = f"{a[1]} {a[2]}"
                    print(ketqua)
    except FileNotFoundError:
        print("Không có đường dẫn")
grade10(r"C:\Users\Admin\Downloads\kiet.csv")
```
