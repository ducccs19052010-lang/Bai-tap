```python
#W11A3
def reverse(path: str) -> int:
    try:
        with open(path, 'r', encoding='utf-8') as file:
            content = file.read().strip()
        reversed_str = content[::-1]
        return int(reversed_str)
    except FileNotFoundError:
        print("Lỗi: Không tìm thấy tệp tin.")
        return 0
    except ValueError:
        print("Lỗi: Nội dung tệp không phải là số nguyên hợp lệ.")
        return 0
file_path = r'C:\Users\Admin\Downloads\kiet1.txt'
result = reverse(file_path)
print(f"Số nguyên đảo ngược là: {result}")
```

```python
#W11A3
def countCharacter(path: str, letter: str) -> int:
    try:
        with open(path, 'r', encoding='utf-8') as file:
            content = file.read()
        content_lower = content.lower()
        letter_lower = letter.lower()
        count = content_lower.count(letter_lower)
        return count
    except FileNotFoundError:
        print("Lỗi: Không tìm thấy tệp tin tại đường dẫn đã cung cấp.")
        return 0
    except Exception as e:
        print(f"Đã xảy ra lỗi: {e}")
        return 0
file_path = r'C:\Users\Admin\Downloads\kiet1.txt'
char_to_find = 'e'
result = countCharacter(file_path, char_to_find)
print(f"Số lần kí tự '{char_to_find}' xuất hiện là: {result}")
```

```python
#W11A8
import re
def solve_fibonacci():
    try:
        with open(r'C:\Users\Admin\Downloads\kiet1.txt', 'r') as f:
            content = f.read()
            match = re.search(r'\d+', content)
            if not match: return
            n = int(match.group())
    except Exception:
        return
    f_list = ['a', 'b']
    for i in range(2, n + 1):
        f_list.append(f_list[i-1] + f_list[i-2])
    for i in range(min(n + 1, len(f_list))):
        print(f"f({i}) = {f_list[i]}")
solve_fibonacci()
```

```python
#W11A10
def getMoney(path):
    try:
        with open(path, 'r') as f:
            content = f.read().replace(',', ' ')
            nums = [int(s) for s in content.split() if s.isdigit()]
    except:
        return []
    return nums
def solve():
    path = r'C:\Users\Admin\Downloads\kiet1.txt'
    nums = getMoney(path)
    if not nums:
        return
    if len(nums) == 1:
        print(nums[0])
        return
    prev2 = 0
    prev1 = 0
    for x in nums:
        current = max(prev1, prev2 + x)
        prev2 = prev1
        prev1 = current
    print(prev1)
solve()
```

```python
#W11A12
def maximumProduct(path):
    try:
        with open(path, 'r') as f:
            lines = f.readlines()
            if len(lines) < 2: return 0
            nums = [int(x) for x in lines[1].split()]
    except:
        return 0
    if len(nums) < 3: return 0
    nums.sort()
    option1 = nums[-1] * nums[-2] * nums[-3]
    option2 = nums[-1] * nums[0] * nums[1]
    return max(option1, option2)
path = r'C:\Users\Admin\Downloads\kiet1.txt'
print(maximumProduct(path))
```

```python
#W11A13
def moveZeroes(fileName):
    try:
        with open(fileName, 'r') as f:
            lines = f.readlines()
            if len(lines) < 2: return []
            nums = [int(x) for x in lines[1].split()]
    except:
        return []
    result = [x for x in nums if x != 0]
    result.extend([0] * (len(nums) - len(result)))
    return result
path = r'C:\Users\Admin\Downloads\kiet1.txt'
print(moveZeroes(path))
```

```python
#W11A14
import csv
def averageTime(path):
    try:
        total_time = 0.0
        count = 0
        with open(path, 'r') as f:
            reader = csv.reader(f)
            next(reader)
            for row in reader:
                if row:
                    total_time += float(row[1])
                    count += 1
        if count == 0: return 0.0
        return round(total_time / count, 2)
    except:
        return 0.0
path = r'C:\Users\Admin\Downloads\kiet1.txt'
print(averageTime(path))
```

```python
#W11A15
def findMovies(path):
    try:
        results = []
        with open(path, 'r', encoding='utf-8') as f:
            for line in f:
                parts = line.strip().split(',')
                if len(parts) >= 3:
                    name = parts[0].strip()
                    rating = float(parts[1].strip())
                    description = parts[2].strip().lower()
                    if rating > 8.0 and "boring" not in description:
                        results.append(name)
        return results
    except:
        return []
path = r'C:\Users\Admin\Downloads\kiet1.txt'
print(findMovies(path))
```

```python
#W11A16
import math
def solve_shapes():
    path = r'C:\Users\Admin\Downloads\kiet1.txt'
    try:
        with open(path, 'r') as f:
            content = f.read()
    except:
        return
    sections = content.split('TYPE:')[1:]
    for section in sections:
        lines = section.strip().split('\n')
        shape_type = lines[0].strip()
        data = ' '.join(lines[1:])
        if shape_type == 'SQUARE':
            side = float(data.split('SIDE:')[1].strip())
            print(f"{side * 4:.2f}")
        elif shape_type == 'RECTANGLE':
            w = float(data.split('WIDTH:')[1].split('HEIGHT:')[0].strip())
            h = float(data.split('HEIGHT:')[1].strip())
            print(f"{(w + h) * 2:.2f}")
        elif shape_type == 'CIRCLE':
            r = float(data.split('RADIUS:')[1].strip())
            print(f"{2 * math.pi * r:.2f}")
solve_shapes()
```

```python
#W11A17
def productExceptSelf(path):
    try:
        with open(path, 'r') as f:
            lines = f.readlines()
            if len(lines) < 2: return []
            nums = [int(x) for x in lines[1].split()]
    except:
        return []
    n = len(nums)
    if n < 2: return nums
    res = [1] * n
    left = 1
    for i in range(n):
        res[i] = left
        left *= nums[i]
    right = 1
    for i in range(n - 1, -1, -1):
        res[i] *= right
        right *= nums[i]
    print(*(res))
    return res
path = r'C:\Users\Admin\Downloads\kiet1.txt'
productExceptSelf(path)
```

```python
#W11A18
import math
def estimatedTime(path):
    try:
        with open(path, 'r') as f:
            content = f.read()
        def get_val(key, text):
            part = text.split(key)[1]
            return float(part.split()[0])
        x1 = get_val('SRC X:', content)
        y1 = get_val('Y:', content.split('SRC X:')[1])
        x2 = get_val('DES X:', content)
        y2 = get_val('Y:', content.split('DES X:')[1])
        v = get_val('VELOCITY:', content)
        dist = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
        print(f"{dist / v:.2f}")
        return round(dist / v, 2)
    except:
        return 0.0
path = r'C:\Users\Admin\Downloads\kiet1.txt'
estimatedTime(path)
```

```python
#W11A20
def imageSmoother(path):
    try:
        with open(path, 'r') as f:
            content = [line.strip() for line in f.readlines() if line.strip()]
        if not content: return
        import re
        m = int(re.search(r'\d+', content[0]).group())
        img = []
        for i in range(1, m + 1):
            row = [int(x) for x in content[i].split()]
            img.append(row)
    except:
        return
    res = [[0] * m for _ in range(m)]
    for r in range(m):
        for c in range(m):
            total, count = 0, 0
            for i in range(r - 1, r + 2):
                for j in range(c - 1, c + 2):
                    if 0 <= i < m and 0 <= j < m:
                        total += img[i][j]
                        count += 1
            res[r][c] = total // count
    for row in res:
        print(*(row))
path = r'C:\Users\Admin\Downloads\kiet1.txt'
imageSmoother(path)
```
