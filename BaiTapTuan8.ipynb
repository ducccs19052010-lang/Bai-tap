```python
import collections
def w7a16():
    # Input: Một dòng chứa các số nguyên
    try:
        arr = list(map(int, input().split()))
    except: return
    
    counts = collections.Counter(arr)
    remove_count = 0
    
    for x, count in counts.items():
        if count > x:
            remove_count += (count - x)
        elif count < x:
            remove_count += count
            
    if remove_count == 0:
        print("YES")
    else:
        print(f"NO {remove_count}")
```

```python
def w7a17():
    # Input: Một dòng chứa các số nguyên
    try:
        arr = list(map(int, input().split()))
    except: return
    
    candidate = None
    count = 0
    for num in arr:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)
    # Không in gì nếu không tìm thấy (theo đề)
```

```python
def w7a18():
    # Input: Một dòng chứa các số nguyên
    try:
        arr = list(map(int, input().split()))
    except: return
    
    evens = sorted([x for x in arr if x % 2 == 0])
    odds = sorted([x for x in arr if x % 2 != 0], reverse=True)
    print(*(evens + odds))
```

```python
def w7a19():
    # Input: Dòng 1 là mảng 1, Dòng 2 là mảng 2
    try:
        arr1 = list(map(int, input().split()))
        arr2 = list(map(int, input().split()))
    except: return
    
    reverse = False
    # Kiểm tra xem input có phải là giảm dần không
    if (len(arr1) > 1 and arr1[0] > arr1[-1]) or (len(arr2) > 1 and arr2[0] > arr2[-1]):
        reverse = True
    
    a = sorted(arr1)
    b = sorted(arr2)
    res = []
    i, j = 0, 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    
    if reverse:
        res = res[::-1]
    print(*res)
```

```python
def w7a20():
    # Input: Dòng 1 là mảng số, Dòng 2 là số k
    try:
        arr = list(map(int, input().split()))
        k = int(input())
    except: return
    
    total = 0
    for num in arr:
        if len(str(abs(num))) == k:
            total += num
    print(total)
```

```python
import collections
def w7a22():
    # Input: Dòng đầu là số N, N dòng tiếp theo là tên
    try:
        line = input().strip()
        if not line: return
        n = int(line)
        full_names = []
        for _ in range(n):
            full_names.append(input().strip())
    except: return

    names = [name.split()[-1] for name in full_names]
    counts = collections.Counter(names)
    max_count = max(counts.values())
    candidates = [name for name, count in counts.items() if count == max_count]
    print(sorted(candidates)[0])
```

```python
import re
def w7a23():
    # Input: Một chuỗi văn bản
    try:
        text = input()
    except: return
    
    words = re.sub(r'[^\w\s]', ' ', text).split()
    count = 0
    for word in words:
        if len(word) >= 2 and word[0].isupper() and word[1:].islower() and word.isalpha():
            count += 1
    print(count)
```

```python
def w7a24():
    # Input: Một chuỗi văn bản
    try:
        text = input()
    except: return
    
    words = text.split()
    if not words: return
    
    res = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            res.append(words[i])
    print(" ".join(res))
```

```python
def w7a25():
    # Input: Dòng 1 là chuỗi số, Dòng 2 là số k
    try:
        number_str = input().strip()
        k = int(input())
    except: return
    
    n = len(number_str)
    if k >= n:
        print(number_str)
        return
        
    to_remove = n - k
    stack = []
    
    for digit in number_str:
        while to_remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            to_remove -= 1
        stack.append(digit)
        
    while to_remove > 0:
        stack.pop()
        to_remove -= 1
        
    print("".join(stack))
```

```python
import re
def w7a26():
    # Input: Một đoạn văn bản
    try:
        text = input()
    except: return
    
    numbers = re.findall(r'\d+', text)
    if not numbers:
        print("0 0")
        return
    
    nums = [int(x) for x in numbers]
    total = sum(nums)
    avg = total / len(nums)
    # Tùy yêu cầu format số thực, ở đây in mặc định
    if avg.is_integer():
        print(f"{total} {int(avg)}")
    else:
        print(f"{total} {avg}")
```

```python
def w7a27():
    # Input: Một chuỗi ngoặc
    try:
        s = input().strip()
    except: return
    
    stack = []
    mapping = {')': '(', ']': '[', '}': '{'}
    
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():
            if not stack or stack.pop() != mapping[char]:
                print("NO")
                return
    
    print("YES" if not stack else "NO")
```

```python
def w7a28():
    # Input: Số nguyên N
    try:
        line = input().strip()
        if not line: return
        N = int(line)
    except: return
    
    def is_composite(val):
        if val <= 3: return False
        for i in range(2, int(val**0.5) + 1):
            if val % i == 0: return True
        return False

    max_seq = []
    curr_seq = []
    
    # Tìm trong khoảng < N
    # Lưu ý: Theo ví dụ đề bài N=10 in 8,9 -> Tức là xét range(1, N)
    for i in range(1, N): 
        if is_composite(i):
            curr_seq.append(i)
        else:
            if len(curr_seq) > len(max_seq):
                max_seq = list(curr_seq)
            curr_seq = []
            
    if len(curr_seq) > len(max_seq):
        max_seq = curr_seq
        
    # In ra dạng chuỗi cách nhau bởi dấu phẩy
    print(", ".join(map(str, max_seq)))
```

```python
def w7a29():
    # Input: Mảng số nguyên
    try:
        arr = list(map(int, input().split()))
    except: return
    
    if not arr: return
    
    max_len = 1
    max_start = 0
    curr_len = 1
    curr_start = 0
    
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            curr_len += 1
        else:
            if curr_len > max_len:
                max_len = curr_len
                max_start = curr_start
            curr_len = 1
            curr_start = i
            
    if curr_len > max_len:
        max_len = curr_len
        max_start = curr_start
        
    res = arr[max_start : max_start + max_len]
    print(*res)
```

```python
def w7a30():
    # Input: Mảng số nguyên
    try:
        arr = list(map(int, input().split()))
    except: return
    
    if not arr: return
    n = len(arr)
    L = [1] * n
    P = [-1] * n
    
    for i in range(n):
        for j in range(i):
            if arr[i] > arr[j] and L[i] < L[j] + 1:
                L[i] = L[j] + 1
                P[i] = j
                
    length = max(L)
    curr = L.index(length)
    lis = []
    while curr != -1:
        lis.append(arr[curr])
        curr = P[curr]
        
    print(*lis[::-1])
```

```python
def w7a31():
    # Input: Chuỗi số
    try:
        s_in = input().strip()
    except: return
    
    s = list(s_in)
    n = len(s)
    
    # Logic: Tìm vị trí i đầu tiên có thể thay thế bằng số nhỏ hơn đứng sau nó
    for i in range(n - 1):
        min_val = s[i]
        min_idx = -1
        
        # Tìm số nhỏ nhất phía sau (nếu có nhiều số bằng nhau thì lấy số ở xa nhất)
        for j in range(i + 1, n):
            if s[j] <= min_val:
                if s[j] < s[i]:
                    min_val = s[j]
                    min_idx = j
        
        if min_idx != -1:
            s[i], s[min_idx] = s[min_idx], s[i]
            # Xử lý số 0 ở đầu (nếu cần in ra số int) -> đề bài ghi là đổi 2 chữ số
            # Theo output mẫu, nếu 10212 -> 1212 (tức là 01212)
            print(int("".join(s)))
            return

    # Nếu không tìm thấy (dãy đã tăng dần), đổi 2 số cuối
    if n >= 2:
        s[-1], s[-2] = s[-2], s[-1]
    
    print(int("".join(s)))
```
