```python
#W7A1
x = list(map(int, input().split()))
target = int(input())
def binary_search(chuoi, target):
    left = 0
    right = len(chuoi) - 1
    while left <= right:
        trungvi = (left + right) // 2
        if chuoi[trungvi] == target:
            return trungvi  
        elif chuoi[trungvi] < target:
            left = trungvi + 1
        else:
            right = trungvi - 1
    return -1  
result = binary_search(x, target)
print(result)
```

```python
#W7A3    Bubble sort
x = list(map(int, input().split()))
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
bubble_sort(x)
print(x)
```

```python
#W7A3   selection sort
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
arr = list(map(int, input().split()))
print(selection_sort(arr))
```

```python
#W7A4
x = list(map(int, input().split()))
ketqua = {}
for i in x:
    if i not in ketqua:
        gtri = x.count(i)
        ketqua[i] = gtri
ketqua1 = dict(sorted(ketqua.items(), key = lambda item: item[1], reverse = True))
ketqua2 = []
for key, value in ketqua1.items():
    ketqua2.extend([key, value])
print(ketqua2)
print(f"{ketqua2[0]} xuat hien nhieu nhat, som nhat, {ketqua2[1]} lan")
```

```python
#W7A4         Dùng thư viện (Cách tối ưu nhất)
x = list(map(int, input().split()))

# Tối ưu việc đếm bằng Dictionary: O(N) thay vì O(N^2)
ketqua = {}
for i in x:
    ketqua[i] = ketqua.get(i, 0) + 1

# Sắp xếp theo số lần xuất hiện giảm dần
sorted_items = sorted(ketqua.items(), key=lambda item: item[1], reverse=True)

# Chuyển về list phẳng (Flatten)
ketqua2 = []
for k, v in sorted_items:
    ketqua2.extend([k, v])

print(ketqua2)
print(f"{ketqua2[0]} xuat hien nhieu nhat, som nhat, {ketqua2[1]} lan")
```
