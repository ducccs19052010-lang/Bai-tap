```python
#W7A1
x = list(map(int, input().split()))
target = int(input())
def binary_search(chuoi, target):
    left = 0
    right = len(chuoi) - 1
    while left <= right:
        trungvi = (left + right) // 2
        if chuoi[trungvi] == target:
            return trungvi  
        elif chuoi[trungvi] < target:
            left = trungvi + 1
        else:
            right = trungvi - 1
    return -1  
result = binary_search(x, target)
print(result)
```

```python
#W7A3    Bubble sort
x = list(map(int, input().split()))
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
bubble_sort(x)
print(x)
```

```python
#W7A3   selection sort
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
arr = list(map(int, input().split()))
print(selection_sort(arr))
```

```python
#W7A4
x = list(map(int, input().split()))
ketqua = {}
for i in x:
    if i not in ketqua:
        gtri = x.count(i)
        ketqua[i] = gtri
ketqua1 = dict(sorted(ketqua.items(), key = lambda item: item[1], reverse = True))
ketqua2 = []
for key, value in ketqua1.items():
    ketqua2.extend([key, value])
print(ketqua2)
print(f"{ketqua2[0]} xuat hien nhieu nhat, som nhat, {ketqua2[1]} lan")
```

```python
#W7A4         Dùng thư viện (Cách tối ưu nhất)
x = list(map(int, input().split()))

# Tối ưu việc đếm bằng Dictionary: O(N) thay vì O(N^2)
ketqua = {}
for i in x:
    ketqua[i] = ketqua.get(i, 0) + 1

# Sắp xếp theo số lần xuất hiện giảm dần
sorted_items = sorted(ketqua.items(), key=lambda item: item[1], reverse=True)

# Chuyển về list phẳng (Flatten)
ketqua2 = []
for k, v in sorted_items:
    ketqua2.extend([k, v])

print(ketqua2)
print(f"{ketqua2[0]} xuat hien nhieu nhat, som nhat, {ketqua2[1]} lan")
```

```python
#W7A5       Cách tối ưu (Hash Map / Frequency Map)
x = list(map(int, input().split()))
k = int(input())
ls1 = {}
count = 0
for num in x:
    sct = k - num
    if sct in ls1:
        count += ls1[sct]
    if num in ls1:
        ls1[num] += 1
    else:
        ls1[num] = 1
print(count)
```

```python
#W7A7           Quy hoạch động (Dynamic Programming)
import ast
x = input()
arr = ast.literal_eval(x)
n = len(arr)
dp = [1] * n
for i in range(1, n):
    for j in range(0, i):
        if arr[i] > arr[j]:
            dp[i] = max(dp[i], dp[j] + 1)
print(max(dp))
```

```python
#W7A8            Xử lí độ phức tạp = O(m x n)
import ast
x = input() 
y = input() 
intervals = ast.literal_eval(x)
queries = ast.literal_eval(y)
ketqua = []
for q in queries:
    min_len = 10*18
    found = False
    for interval in intervals:
        start = interval[0]
        end = interval[1]
        if start <= q <= end:
            current_len = end - start + 1
            if current_len < min_len:
                min_len = current_len
            found = True
    ketqua.append(min_len if found else -1)
print(ketqua)
```

```python
#W7A8         Xử lí độ phức tạp O(nlogn + mlogm) (Máy quét (Sweep line) và Hàng đợi ưu tiên (Min-Heap))
import ast
import heapq
x = input()
y = input()
intervals = ast.literal_eval(x)
queries = ast.literal_eval(y)
def min_interval_length(intervals, queries):
    intervals.sort(key = lambda x: x[0])             #lambda x là 1 hàm ẩn danh, có thể hiểu x đại diện cho 1 khoảng trong danh sách
    sorted_queries = sorted((q, i) for i, q in enumerate(queries))
    min_heap = []           #Lưu trữ (Độ dài, điểm kết thúc r)
    result = {}             #Lưu trữ kết quả theo index gốc: {index: độ dài}
    i = 0
    n = len(intervals)
    for q_val, q_index in sorted_queries:
        while i < n and intervals[i][0] <= q_val:
            l, r = intervals[i]
            length = r - l + 1
            heapq.heappush(min_heap, (length, r))           #Sắp xếp đưa phần tử nhỏ nhất lên đầu và đưa vào min_heap
            i += 1
        while min_heap and min_heap[0][1] < q_val:
            heapq.heappop(min_heap)
        if min_heap:
            result[q_index] = min_heap[0][0]
        else:
            result[q_index] = -1
    return [result[i] for i in range(len(queries))]
print(min_interval_length(intervals, queries))
```

```python
#W7A10
import ast
x = input()
ls = ast.literal_eval(x)
k = int(input())
tongcon = []
n = len(ls)
#Prefix Sum
p = [0] * (n + 1)
for i in range(n):
    p[i + 1] = p[i] + ls[i]
for i in range(n):        
    for j in range(i, n):  
        current_sum = p[j + 1] - p[i]
        tongcon.append(current_sum)
ketqua = 0
for i in range(len(tongcon)):
    n = tongcon[i] % k
    ketqua = max(ketqua, n)
print(ketqua)
```

```python
#W7A13
x = input().split()
y = [int(i) for i in x]
ls = sorted(y)
ketqua = 0
left = 0
for right in range(len(ls)):
    while ls[right] - ls[left] > 1:
        left += 1
    if ls[right] - ls[left] == 1:
        ketqua = max(ketqua, right - left + 1)
print(ketqua)
```
