```python
#P060
import math
n = int(input())
def kiemtrasnt(x):
    if x < 2:
        return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return False
    return True
if n == 0 or n == 1:
    print(0)
else:
    for i in range(n, 1, -1):
        if kiemtrasnt(i):
            print(i)
            break
```
