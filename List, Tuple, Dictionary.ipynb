```python
A = [1, 2, 3, 4, 5, 6]
A.append("kiet") #Them 'kiet' vao cuoi list
print(A)
```

```python
A = [1, 2, 3, 4, 5, 6]
A.insert(3,"kiet") #Them 'kiet' vao index 3 cua list
print(A)
```

```python
A = [1, 2, 3, 4, 5, 6]
A.extend([8, 9]) #Noi list [8, 9] vao cuoi list
print(A)
```

```python
A = [1, 2, 3, 4, 5, 6]
A.remove(2) #Xoa phan tu dau tien co gia tri la 2
print(A)
```

```python
A = [1, 2, 3, 4, 5, 6]
A.pop(3) #Xoa va tra ve phan tu tai index 3 (Mac dinh la cuoi)
print(A)
```

```python
A = [1, 2, 3, 4, 5, 6]
A.clear() #Xoa toan bo phan tu cua list
print(A)
```

```python
A = [3, 5, 6, 2, 1, 4]
A.sort() #Sap xep list tai cho (Thay doi list goc)
print(A)
```

```python
A = [1, 2, 3, 3, 5, 6]
i = A.index(3) #Tim vi tri index dau tien cua gia tri 3
print(i)
```

```python
A = [1, 2, 3, 3, 3, 6]
i = A.count(3) #Dem so lan xuat hien cua gia tri 3
print(i)
```
