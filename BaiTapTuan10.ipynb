```python
def solve():
    try:
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
    except ValueError:
        return

    res = []
    i = 0
    j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    print(*res)

solve()
```

```python
def solve():
    s = list(input().strip())
    n = len(s)
    
    for i in range(n - 1):
        min_val = s[i]
        min_idx = -1
        
        for j in range(i + 1, n):
            if s[j] < min_val:
                min_val = s[j]
                min_idx = j
            elif s[j] == min_val and min_idx != -1:
                min_idx = j
                
        if min_idx != -1:
            if i == 0 and min_val == '0' and n > 1:
                continue 
            s[i], s[min_idx] = s[min_idx], s[i]
            print("".join(s))
            return

    s[-1], s[-2] = s[-2], s[-1]
    print("".join(s))

solve()
```

```python
def check(a, b):
    n, m = len(a), len(b)
    dp = [[False] * (m + 1) for _ in range(n + 1)]
    dp[0][0] = True
    
    for i in range(n):
        for j in range(m + 1):
            if dp[i][j]:
                if j < m and a[i].upper() == b[j]:
                    dp[i + 1][j + 1] = True
                if a[i].islower():
                    dp[i + 1][j] = True
                    
    return "YES" if dp[n][m] else "NO"

def solve():
    try:
        q_line = input().strip()
        if not q_line: return
        Q = int(q_line)
        for _ in range(Q):
            a = input().strip()
            b = input().strip()
            print(check(a, b))
    except:
        pass

solve()
```

```python
def solve():
    try:
        l1 = list(map(int, input().split()))
        l2 = list(map(int, input().split()))
    except:
        return

    n = len(l1)
    m = len(l2)
    
    if n == 0:
        print("YES")
        return
    if n > m:
        print("NO")
        return

    found = False
    for i in range(m - n + 1):
        if l2[i:i+n] == l1:
            found = True
            break
            
    print("YES" if found else "NO")

solve()
```

```python
def solve():
    try:
        input() 
        nums = list(map(int, input().split()))
    except:
        return
        
    i = len(nums) - 2
    while i >= 0 and nums[i] >= nums[i + 1]:
        i -= 1
        
    if i >= 0:
        j = len(nums) - 1
        while nums[j] <= nums[i]:
            j -= 1
        nums[i], nums[j] = nums[j], nums[i]
        
    left, right = i + 1, len(nums) - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
        
    print(*nums)

solve()
```

```python
def canPlaceFlowers(flowerbed, k):
    count = 0
    flowerbed = [0] + flowerbed + [0]
    for i in range(1, len(flowerbed) - 1):
        if flowerbed[i-1] == 0 and flowerbed[i] == 0 and flowerbed[i+1] == 0:
            flowerbed[i] = 1
            count += 1
    return count >= k

def solve():
    try:
        f_in = input().strip()
        if not f_in: return
        flowerbed = [int(x) for x in f_in]
        k = int(input())
        print(canPlaceFlowers(flowerbed, k))
    except:
        pass

solve()
```

```python
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}
    
    for move in moves:
        if move == 'U': y += 1
        elif move == 'D': y -= 1
        elif move == 'R': x += 1
        elif move == 'L': x -= 1
        
        if (x, y) in visited:
            return True
        visited.add((x, y))
        
    return False

def solve():
    moves = input().strip()
    print(is_path_crossing(moves))

solve()
```

```python
def Median(nums):
    A = list(nums)
    n = len(A)
    for i in range(n):
        for j in range(0, n-i-1):
            if A[j] > A[j+1]:
                A[j], A[j+1] = A[j+1], A[j]
    return A[2]

def solve():
    try:
        inp = input().strip()
        nums = list(map(int, inp.split()))
        if len(nums) != 5: # Fallback if input format differs
             nums = [int(x) for x in inp]
        print(Median(nums))
    except:
        pass

solve()
```

```python
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    radius = 0
    
    for house in houses:
        idx = bisect.bisect_left(heaters, house)
        
        dist1 = float('inf')
        dist2 = float('inf')
        
        if idx < len(heaters):
            dist1 = heaters[idx] - house
        if idx > 0:
            dist2 = house - heaters[idx - 1]
            
        radius = max(radius, min(dist1, dist2))
        
    return radius

def solve():
    try:
        houses = list(map(int, input().split()))
        heaters = list(map(int, input().split()))
        print(findRadius(houses, heaters))
    except:
        pass

solve()
```
