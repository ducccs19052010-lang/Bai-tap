```python
x = float(input("Nhập số thực: "))
xuong = int(x) if x >= 0 else int(x) - 1 if x != int(x) else int(x)
len_ = int(x) + 1 if x > int(x) else int(x)
gan_nhat = int(x + 0.5) if x >= 0 else int(x - 0.5)
print(f"{len_} {xuong} {gan_nhat}")
```
