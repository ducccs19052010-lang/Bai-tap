```python
#W6A1
chuoi = list(map(int, input().split()))
chuoi1 = []
for so in chuoi:
    if so not in chuoi1:
        chuoi1.append(so)
print(chuoi1)
```

```python
#W6A2
chuoi = list(map(int, input().split()))
tichluy = []
def tichluy1(x):
    tong = 0
    for i in range(0, x + 1):
        tong = tong + chuoi[i]
    return tong
for i in range(0, len(chuoi)):
    tichluy.append(tichluy1(i))
print(tichluy)
```

```python
#W6A3
chuoi = tuple(map(int, input().split()))
k = int(input())
s = len(chuoi)
if k > s:
    h = k - s
    arr = chuoi[:h]               
    arr2 = chuoi[h:]
    print(arr2 + arr)
else:
    arr = chuoi[:k]
    arr2 = chuoi[k:]
    print(arr2 + arr)

```

```python
#W6A4
chuoi = input()
ketqua = {}
for cap in chuoi.split():
    if ':' in cap:
        key, value = cap.split(':', 1) #Cu phap: split(<Ki tu hoac chuoi dung phan tach>, <so lan toi da duoc phep phan tach>)
        if key in ketqua:
            ketqua[key].append(value)
        else:
            ketqua[key] = [value]
print(ketqua)
for key in ketqua:
    print(key)
```
