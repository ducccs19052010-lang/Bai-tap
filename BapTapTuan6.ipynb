```python
#W6A2
chuoi = list(map(int, input().split()))
tichluy = []
def tichluy1(x):
    tong = 0
    for i in range(0, x + 1):
        tong = tong + chuoi[i]
    return tong
for i in range(0, len(chuoi)):
    tichluy.append(tichluy1(i))
print(tichluy)
```

```python
#W6A3
chuoi = tuple(map(int, input().split()))
k = int(input())
s = len(chuoi)
if k > s:
    h = k - s
    arr = chuoi[:h]               
    arr2 = chuoi[h:]
    print(arr2 + arr)
else:
    arr = chuoi[:k]
    arr2 = chuoi[k:]
    print(tuple(arr2 + arr))

```

```python
#W6A4
chuoi = input()
ketqua = {}
for cap in chuoi.split():
    if ':' in cap:
        key, value = cap.split(':', 1) #Cu phap: split(<Ki tu hoac chuoi dung phan tach>, <so lan toi da duoc phep phan tach>)
        if key in ketqua:
            ketqua[key].append(value)
        else:
            ketqua[key] = [value]
print(ketqua)
for key in ketqua:
    print(key)
```

```python
#W6A5
ls = {'positives':0, 'negatives':0, 'zeros':0}
arr = list(map(int, input().split()))
for i in range(len(arr)):
    if arr[i] == 0:
        ls['zeros'] += 1
    if arr[i] > 0:
        ls['positives'] += 1
    if arr[i] < 0:
        ls['negatives'] += 1
print(ls)
```

```python
#W6A9 cach standard
import ast
x1 = input()
x2 = input()
dx1 = ast.literal_eval(x1)
dx2 = ast.literal_eval(x2)
result = dx1.copy()
for key, value in dx2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value
ketqua = dict(sorted(result.items()))
print(ketqua)
```

```python
#W6A10
x = input()
k = int(input())
ls = [int(num) for num in x.split()]              # .split() de xoa cac dau cach co trong chuoi
ketqua = set()
n = len(ls)
for i in range(n):
    for j in range(i + 1, n):
        if ls[i] + ls[j] == k:
            x = (ls[i],ls[j])
            ketqua.add(x)
arr = sorted(list(ketqua))
print(arr)
```

```python
#W4A12
x = input()
ls = [int(num) for num in x.split()]
dic = {}
for num in ls:
    if num in dic:
        dic[num] += 1
    else:
        dic[num] = 1
gtln = -1
ketqua = num
for num, fre in dic.items():
    if fre > gtln:
        gtln = fre
        ketqua = num
    elif fre == gtln:
        if num < ketqua:
            ketqua = num
print(ketqua)
```
