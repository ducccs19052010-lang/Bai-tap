NEXT DAY

```python
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    def is_leap_year(self):
        #Năm nhuận là năm chia hết cho 400 hoặc chia hết cho 4 nhưng không chia hết cho 100
        return (self.year % 400 == 0) or (self.year % 4 == 0 and self.year % 100 != 0)
    def days_in_month(self):
        if self.month in [1, 3, 5, 7, 8, 10,12]:
            return 31
        elif self.month in [4, 6, 9, 11]:
            return 30
        elif self.month == 2:
            return 29 if self.is_leap_year() else 28
        else:
            return 0 #Tháng không hợp lệ
    def is_valid(self):
        if self.year < 0:
            return False
        if self.month < 1 or self.month > 12:
            return False
        max_days = self.days_in_month()
        if self.day < 1 or self.day > max_days:
            return False
        return True
    def next_day(self):
        self.day += 1
        if self.day > self.days_in_month():
            self.day = 1
            self.month += 1
        if self.month > 12:
            self.month = 1
            self.year += 1
    def print_date(self):
        print(f"{self.day:02d}/{self.month:02d}/{self.year}")
def main():
    try:
        input_string = input()
        parts = input_string.split('/')
        if len(parts) != 3:
            print("INVALID")
            return
        d, m, y = int(parts[0]), int(parts[1]), int(parts[2])
        my_date = Date(d, m, y)
        if my_date.is_valid():
            my_date.next_day()
            my_date.print_date()
        else:
            print("INVALID")
    except ValueError:
        print("INVALID")
if __name__ == "__main__":
        main()
```

TỌA ĐỘ

```python
import math
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def gtd(self):
        if self.x == 0 and self.y == 0:
            return("la goc toa do")
        else:
            return("Khong la goc toa do")
    def kiemtra(self):
        if self.x == 0 and self.y != 0:
            return("Nam tren truc tung")
        elif self.x != 0 and self.y == 0:
            return("Nam tren truc hoanh")
        else:
            return("Khong thuoc ca truc tung va truc hoanh")
    def distance(self):
        x = math.sqrt(self.x**2 + self.y**2)
        return(f"{x:.2f}")
def main():
    try:
        x, y = map(int, input().split())
        ketqua = Point(x,y)
        print(ketqua.gtd())
        print(ketqua.kiemtra())
        print(ketqua.distance())
    except ValueError:
        print("INVALID")
if __name__ == "__main__":
    main()
```

HỆ THỐNG MÔ PHỎNG TRẬN ĐẤU ROBOT!

```python
class Robot:
    def __init__(self, name, hp, atk):
        self.name = name
        self.hp = hp
        self.atk = atk
        self.max_hp = hp
    def is_alive(self):
        if self.hp > 0:
            return True
        else:
            return False
    def receive_damage(self, amount):
        self.hp -= amount
        if self.hp < 0:
            self.hp = 0
    def attack(self, other):
        other.receive_damage(self.atk)
class Tanker(Robot):
    def __init__(self, name, hp, atk, defense):
        super().__init__(name, hp, atk)
        self.defense = defense
        self.shield = 0
    def attack(self, other):
        super().attack(other)
        self.shield += 1
    def receive_damage(self, amount):
        actual_damage = amount - self.defense - self.shield  
        if actual_damage < 0:
            actual_damage = 0
        super().receive_damage(actual_damage)
        if self.shield > 0:
            self.shield = 0
class Assassin(Robot):
    def __init__(self, name, hp, atk):
        super().__init__(name, hp, atk)
        self.crit_count = 0
        self.stealth = False
    def attack(self, other):
        self.crit_count += 1
        damage = self.atk
        if self.crit_count == 3:
            damage = self.atk * 2
            self.crit_count = 0
        other.receive_damage(damage)
    def receive_damage(self, amount):
        if self.stealth:
            amount = amount // 2
        super().receive_damage(amount)
        if not self.stealth and self.hp <= self.max_hp * 0.3:
            self.stealth = True
def solve():
    line1 = input().split()
    tank_name = line1[1]
    tank_hp = int(line1[2])
    tank_atk = int(line1[3])
    tank_def = int(line1[4])
    tanker = Tanker(tank_name, tank_hp, tank_atk, tank_def)
    line2 = input().split()
    asn_name = line2[1]
    asn_hp = int(line2[2])
    asn_atk = int(line2[3])
    assassin = Assassin(asn_name, asn_hp, asn_atk)
    n = int(input())
    for _ in range(n):
        attack_name = input().strip()    #.strip() được dùng để loại bỏ các ký tự thừa (mặc định là khoảng trắng, dấu cách, dấu tab, xuống dòng) ở hai đầu của một chuỗi.
        if attack_name == tanker.name:
            if tanker.is_alive() and assassin.is_alive():
                tanker.attack(assassin)
        elif attack_name == assassin.name:
            if tanker.is_alive() and assassin.is_alive():
                 assassin.attack(tanker)
    print(f"{tanker.name}: HP={tanker.hp}")
    print(f"{assassin.name}: HP={assassin.hp}")
if __name__ == "__main__":
    solve()
```

SỐ PHỨC

```python
import math
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __add__(self, other):
        new_real = self.real + other.real
        new_imag = self.imag + other.imag
        return ComplexNumber(new_real, new_imag)
    def __mul__(self, other):
        new_real = self.real * other.real - self.imag * other.imag
        new_imag = self.real * other.imag + other.real * self.imag
        return ComplexNumber(new_real, new_imag)
    def __abs__(self):
        return math.sqrt(self.real**2 + self.imag**2)
    def conjugate(self):
        return ComplexNumber(self.real, -self.imag)
    def __str__(self):
        dau = "+" if self.imag >=0 else "-"
        return f"{self.real:.2f} {dau} {abs(self.imag):.2f}i"
def solve():
    line1 = input().split()
    z1 = ComplexNumber(float(line1[0]), float(line1[1]))
    line2 = input().split()
    z2 = ComplexNumber(float(line2[0]), float(line2[1]))
    tong = z1 + z2
    print(tong)
    tich = z1 * z2
    print(tich)
    module = abs(tich)
    print(f"{module:.2f}")
    lh = tich.conjugate()
    print(lh)
if __name__ == "__main__":
    solve()
```

LỚP TAM GIÁC

```python
import math
class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def is_valid(self):
        if self.a + self.b > self.c and self.a + self.c > self.b and self.b + self.c > self.a and self.a > 0 and self.b > 0 and self.c > 0:
            return True
        else:
            return False
    def getArea(self):
        if not self.is_valid():
            return -1
        else:
            C = (self.a + self.b + self.c)/2
            S = math.sqrt(C * (C - self.a) * (C - self.b) * (C - self.c))
            return S
def solve():
    a, b, c = map(float, input().split())
    ketqua = Triangle(a, b, c)
    area = ketqua.getArea()
    if area == -1:
        print("invalid")
    else:
        print(f"{area:.2f}")
if __name__ == "__main__":
    solve()
```

XỬ LÍ THỜI GIAN (TIMESTAMP)

```python
class TimeStamp:
    def __init__(self, hours, minutes, seconds):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds
    def isValid(self):
        if self.hours < 0 or self.hours >= 24:
            return False
        if self.minutes < 0 or self.minutes >= 60:
            return False
        if self.seconds < 0 or self.seconds >= 60:
            return False
        return True
    def nextSecond(self):
        if self.seconds != 59:
            self.seconds += 1
        else:
            self.seconds = 00
            if self.minutes != 59:
                self.minutes += 1
            else:
                self.minutes = 00
                if self.hours != 23:
                    self.hours += 1
                else:
                    self.hours = 00
    def previousSecond(self):
        if self.seconds != 0:
            self.seconds -= 1
        else:
            self.seconds = 59
            if self.minutes != 0:
                self.minutes -= 1
            else:
                self.minutes = 59
                if self.hours != 0:
                    self.hours -= 1
                else:
                    self.hours = 23
    def __str__(self):
        return f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}"
def solve():
    line = input().split(":")
    h, m, s = map(int, line)
    t = TimeStamp(h, m ,s)
    if t.isValid():
        t.nextSecond()
        print(t)
        t.previousSecond()
        t.previousSecond()
        print(t)
    else:
        print("INVALID")
if __name__ == "__main__":
    solve()
    
```

PHÂN SỐ

```python
import math
class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Mẫu số không thể bằng 0")
        if den < 0:
            num = -num
            den = -den
        ucln = math.gcd(num, den)
        self.num = num // ucln
        self.den = den // ucln

    def __add__(self, other):
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def __sub__(self, other):
        new_num = self.num * other.den - self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def __mul__(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def __truediv__(self, other):
        if other.num == 0:
            raise ValueError("Không thể chia cho phân số 0")
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)

    def __str__(self):
        if self.den == 1:
            return str(self.num)
        else:
            return f"{self.num}/{self.den}"

def solve():
    try:
        line = input().split()
        if len(line) != 5:
            print("Nhập sai định dạng")
            return
        a = int(line[0]) 
        b = int(line[1])
        op = line[2]
        c = int(line[3])
        d = int(line[4])
        f1 = Fraction(a, b)
        f2 = Fraction(c, d)
        result = None
        if op == "+":
            result = f1 + f2
        elif op == "-":
            result = f1 - f2
        elif op == "*":
            result = f1 * f2
        elif op == "/":
            result = f1 / f2
        if result is not None:
            print(result)
        else:
            print("Toán tử không hợp lệ")

    except Exception as e:
        print(f"Lỗi: {e}")

if __name__ == "__main__":
    solve()
```

TÀI KHOẢN NGÂN HÀNG

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            return
    def __str__(self):
        return f"{self.balance:.2f}" 
def solve():
    line = input().split()
    n = int(input())
    owner = line[0]
    balance = float(line[1])
    ketqua = BankAccount(owner, balance)
    for _ in range(n):
        x = input().split()
        if x[0] == "DEPOSIT":
            ketqua.deposit(float(x[1]))
        else:
            ketqua.withdraw(float(x[1]))
    print(ketqua)
if __name__ == "__main__":
    solve()        

```
