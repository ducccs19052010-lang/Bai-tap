```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
        self.pi = 3.14

    def surface_area(self):
        # Diện tích toàn phần = 2*pi*r*h + 2*pi*r^2
        return 2 * self.pi * self.r * self.h + 2 * self.pi * (self.r ** 2)

    def volume(self):
        # Thể tích = pi*r^2*h
        return self.pi * (self.r ** 2) * self.h

try:
    line = input().split()
    if len(line) == 2:
        r, h = map(float, line)
        cyl = Cylinder(r, h)
        print(f"{cyl.surface_area():.2f} {cyl.volume():.2f}")
except:
    pass
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def check_position(self):
        if self.x == 0 and self.y == 0:
            print("M la goc toa do")
        elif self.x == 0:
            print("M nam tren truc tung")
        elif self.y == 0:
            print("M nam tren truc hoanh")
        else:
            print("M khong nam tren truc toa do")

    def distance_to_origin(self):
        return math.sqrt(self.x**2 + self.y**2)

try:
    line = input().split()
    x, y = float(line[0]), float(line[1])
    p = Point(x, y)
    p.check_position()
    print(f"{p.distance_to_origin():.2f}")
except:
    pass
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        self.items.append({'name': name, 'price': price})

    def remove_item(self, name):
        self.items = [item for item in self.items if item['name'] != name]

    def is_empty(self):
        return len(self.items) == 0

    def get_total(self):
        return sum(item['price'] for item in self.items)

    def display(self):
        for item in self.items:
            print(f"{item['name']}: {item['price']}")

    def clear(self):
        self.items = []

cart = ShoppingCart()
while True:
    try:
        cmd = input()
        if cmd.lower() == 'yes':
            break
        # Logic xử lý menu chi tiết
    except:
        break
```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w *= k
        self.h *= k

try:
    line = input().split()
    w, h, k = float(line[0]), float(line[1]), float(line[2])
    rect = Rectangle(w, h)
    rect.scale(k)
    print(f"{rect.area():.2f} {rect.perimeter():.2f}")
except:
    pass
```

```python
import math

class Fraction:
    def __init__(self, num, den):
        if den == 0: raise ValueError
        common = math.gcd(num, den)
        self.num = num // common
        self.den = den // common
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den

    def __str__(self):
        return f"{self.num}/{self.den}" if self.den != 1 else f"{self.num}"

    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)

    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)

    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)

    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)

try:
    line = input().split() # a b op c d
    a, b = int(line[0]), int(line[1])
    op = line[2]
    c, d = int(line[3]), int(line[4])

    f1 = Fraction(a, b)
    f2 = Fraction(c, d)

    if op == '+': print(f1.add(f2))
    elif op == '-': print(f1.sub(f2))
    elif op == '*': print(f1.mul(f2))
    elif op == '/': print(f1.div(f2))
except:
    pass
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"

try:
    line1 = input().rsplit(' ', 1)
    owner = line1[0]
    balance = float(line1[1])
    acc = BankAccount(owner, balance)
    
    n = int(input())
    for _ in range(n):
        cmd_line = input().split()
        op = cmd_line[0]
        amt = float(cmd_line[1])
        if op == 'DEPOSIT':
            acc.deposit(amt)
        elif op == 'WITHDRAW':
            acc.withdraw(amt)
            
    print(f"{acc.balance:.2f}")
except:
    pass
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, score):
        self.scores.append(score)

    def avg(self):
        if not self.scores: return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        a = self.avg()
        if a >= 8.0: return "Excellent"
        elif a >= 6.5: return "Good"
        elif a >= 5.0: return "Average"
        else: return "Poor"

try:
    name = input()
    n = int(input())
    st = Student(name)
    for _ in range(n):
        line = input().rsplit(' ', 1)
        st.add_score(float(line[1]))
        
    print(f"{st.name} {st.avg():.2f} {st.rank()}")
except:
    pass
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = int(year)

class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_by_author(self, name):
        return sum(1 for b in self.books if b.author == name)

    def count_by_year(self, year):
        return sum(1 for b in self.books if b.year == year)

lib = Library()
try:
    n = int(input())
    for _ in range(n):
        line = input().strip()
        if line.startswith("ADD"):
            # ADD title; author; year
            parts = line[4:].split(';')
            lib.add(Book(parts[0].strip(), parts[1].strip(), parts[2].strip()))
        elif line.startswith("COUNTYEAR"):
            y = int(line.split()[1])
            print(lib.count_by_year(y))
        elif line.startswith("COUNT"):
            # COUNT author
            auth = line[6:].strip()
            print(lib.count_by_author(auth))
except:
    pass
```

```python
import math

class Vector3D:
    def __init__(self, x, y, z):
        self.x, self.y, self.z = x, y, z
    
    def sub(self, other):
        return Vector3D(self.x - other.x, self.y - other.y, self.z - other.z)
    
    def cross(self, other):
        return Vector3D(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )
    
    def dot(self, other):
        return self.x * other.x + self.y * other.y + self.z * other.z
    
    def mag(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)

def get_point():
    line = list(map(float, input().split()))
    return Vector3D(line[0], line[1], line[2])

try:
    A = get_point()
    B = get_point()
    C = get_point()
    D = get_point()

    # Normal of ABC: (B-A) x (C-A)
    AB = B.sub(A)
    AC = C.sub(A)
    n1 = AB.cross(AC)

    # Normal of BCD: (C-B) x (D-B)
    BC = C.sub(B)
    BD = D.sub(B)
    n2 = BC.cross(BD)

    # Angle between normals
    numerator = abs(n1.dot(n2))
    denominator = n1.mag() * n2.mag()
    
    if denominator == 0:
        print("0.00")
    else:
        cos_phi = numerator / denominator
        # Clamp value for float errors
        cos_phi = max(min(cos_phi, 1.0), -1.0)
        angle_rad = math.acos(cos_phi)
        print(f"{math.degrees(angle_rad):.2f}")
except:
    pass
```

```python
class ComplexNumber:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def multiply(self, other):
        # (a+bi)(c+di) = (ac - bd) + (ad + bc)i
        r = self.real * other.real - self.img * other.img
        i = self.real * other.img + self.img * other.real
        return ComplexNumber(r, i)

try:
    l1 = list(map(float, input().split()))
    l2 = list(map(float, input().split()))
    
    c1 = ComplexNumber(l1[0], l1[1])
    c2 = ComplexNumber(l2[0], l2[1])
    
    res = c1.multiply(c2)
    
    # Định dạng output số nguyên nếu không có phần thập phân để khớp mẫu
    r_print = int(res.real) if res.real.is_integer() else res.real
    i_print = int(res.img) if res.img.is_integer() else res.img
    print(f"{r_print} {i_print}")
except:
    pass
```

```python
import math

class Triangle:
    def __init__(self, e1, e2, e3):
        self.e1 = e1
        self.e2 = e2
        self.e3 = e3

    def get_area(self):
        a, b, c = self.e1, self.e2, self.e3
        if a + b > c and a + c > b and b + c > a:
            p = (a + b + c) / 2
            return math.sqrt(p * (p - a) * (p - b) * (p - c))
        return -1

try:
    line = list(map(int, input().split()))
    t = Triangle(line[0], line[1], line[2])
    area = t.get_area()
    if area == -1:
        print("invalid")
    else:
        print(f"{area:.2f}")
except:
    pass
```
